```python

# ==========================================
# 1. IMPORT REQUIRED LIBRARIES
# ==========================================

from llama_index.core.prompts import ChatPromptTemplate
from llama_index.core.llms import ChatMessage
from llama_index.llms.ollama import Ollama


# ==========================================
# 2. INITIALIZE DEEPSEEK R1
# ==========================================

llm = Ollama(
    model="deepseek-r1:latest",
    request_timeout=120.0,
    temperature=0.1,
    context_window=1024,
    additional_kwargs={"num_predict": 256}
)


# ==========================================
# 3. CREATE CHAT PROMPT TEMPLATE
# ==========================================

template = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful chatbot. Be concise and accurate."),

    ("user", """
    Previous Conversation:
    {chat_history}

    Current User Query:
    {user_query}
    """)
])


# ==========================================
# 4. DEFINE CHATBOT FUNCTION
# ==========================================

def chat():

    chat_history = []

    print("\nLlamaIndex DeepSeek R1 Chatbot")
    print("Type 'exit' to end the conversation.\n")

    while True:

        user_query = input("You: ")

        if user_query.lower().strip() == "exit":
            print("\nChatbot: Goodbye!")
            break

        try:

            # Convert chat history into text
            history_text = "\n".join(
                f"{role}: {message}"
                for role, message in chat_history
            )

            # Format the prompt
            prompt_messages = template.format_messages(
                chat_history=history_text,
                user_query=user_query
            )

            # Generate response using DeepSeek R1
            response = llm.chat(prompt_messages)

            answer = response.message.content

            print("\nDeepSeek R1:", answer, "\n")

            # Update conversation history
            chat_history.append(("Human", user_query))
            chat_history.append(("Assistant", answer))

        except Exception as e:

            print("Error:", e)


# ==========================================
# 5. RUN THE CHATBOT
# ==========================================

chat()
```
